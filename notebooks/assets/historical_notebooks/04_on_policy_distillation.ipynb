## 1. Understand the experiment

Think of a student solving a math problem on paper. In 03, we supplied a worked
solution and taught the student to predict its words. Here, the student writes
its **own** solution. After each partial answer, the teacher tells us how likely
each possible next token should be. A token is a small piece of text, sometimes
a whole word and sometimes only part of one.

The teacher is **frozen**, so its weights never change. Only the small student learns.
The teacher's probabilities are guidance about text, not a guarantee that the
arithmetic is correct. A teacher can still prefer a wrong continuation.

| Starting point | Additional practice | Purpose |
| :--- | :--- | :--- |
| **Starting SFT** | None | Measure the checkpoint we already have |
| **Continued SFT** | 1,024 new reference solutions | Check what more ordinary supervised practice achieves |
| **On-policy KD** | The same 1,024 questions, student-written responses, teacher distributions | Test learning on prefixes the student actually visits |

The full training recipe starts both arms from **identical checkpoint bytes**, uses
the same question order, and makes 128 optimizer updates with eight examples per
update. The next EOS comparison stops each fresh KD pilot after at most 32 updates. Their token
counts, compute cost, and learning signals differ. This comparison measures the two methods together with their different learning
signals and token budgets. Isolating the sampling policy would require another arm.

![Three arms from the same checkpoint](assets/04_on_policy_distillation/controlled_comparison.png)

### How this relates to the paper

We implement the fully on-policy case of [GKD, Section 3.1](https://arxiv.org/html/2306.13649v3#S3.SS1):
sample from the current student, hold those token IDs fixed, and match the teacher's
next-token distributions on those prefixes. We use **forward KL** and teacher
temperature `0.1`, following its arithmetic experiment choices. Our Qwen models,
reference-solution warm start, prompt, data budget, and training scale are different.
This is an educational adaptation, not the paper's reported benchmark experiment.

**On-policy** describes where responses come from, not the direction of KL.
01 used reverse KL to inspect mechanics. 04 uses forward KL for this experiment.
Teacher-generated answers are used only to measure teacher validation performance.
Training targets come from teacher distributions on student-written prefixes.
This experiment uses full student fine-tuning with no answer filtering.

This single probe supports a teacher-context explanation. It does not establish
that a corrected training run will improve accuracy. The example response was

separate measurements.

### The safeguard gap in the earlier code

The earlier guard paused only when 24 of the latest 32 sampled responses were
capped. A reference-loss increase produced a warning, and health was checked
before the newest validation measurements were recorded. The grouped pilot could
therefore continue while its answer quality deteriorated.

The revised policy checks the newest validation before another update. It pauses
when reference NLL rises over 10%, strict accuracy falls over five percentage
points, or more than 8 of the latest 32 responses hit the token limit. Each
threshold is an engineering review setting. Its configuration is signed into the
new experiment and saved with recovery metadata.

### Where the deterioration first appears

![Verified validation history and sampled-response caps](assets/04_on_policy_distillation/pilot_history.png)

*Points show saved GPU measurements. The dotted line marks update 8, the first

At update 8, raw-token strict matches fell from 33/64 to 17/64. Grouped STOP fell
to 29/64, with five previously wrong answers becoming correct and nine previously
correct answers becoming wrong. Reference NLL was still close to baseline at
0.679 and 0.673, respectively. A guard that only watches NLL would miss this early
accuracy decline. The selected accuracy limit requests a pause at update 8 in both arms.

One grouped-stop regression changes a correct staircase calculation of 15 + 9 = 24
into an invented extra staircase term and concludes 25. Another changes a correct
40-mile route into 20 miles. Both use the requested final-answer marker. Arithmetic
and format failures both occur in the recorded responses.

The fixed-prefix grouped KL falls from 0.918 to about 0.646 by the grouped arm's

The contribution of teacher sharpness, update size, and student capacity requires
a controlled follow-up. New probes record entropy and ordinary/EOS loss components
so those hypotheses can be examined directly.

### 3.1 What exactly receives a loss?

`input_ids` contains prompt tokens followed by response tokens. For SFT, the
response comes from the reference solution. For on-policy KD, it comes from the
student. Prompt positions provide context but have no direct training loss. Processing
that context still affects how the student predicts its response.

We process one unpadded response at a time during distillation. This removes the
need to guess which repeated end-token IDs are padding. SFT still uses explicit
`-100` labels to ignore prompt/padding positions. A real end token remains a target.

![The one-token prediction relationship](assets/04_on_policy_distillation/token_alignment.png)

If the prompt has `P` tokens, the first response token is at complete-sequence
index `P`, but its prediction is at logits index `P - 1`. The last prompt token's
logits predict what comes next. We therefore keep `logits[:, P - 1:-1, :]`.
The final logits position predicts an unobserved token and is excluded.


```python
# A real reference example makes the boundary visible before any model is loaded.
example = train_examples[0]
prompt_length = example["prompt_length"]
preview_batch = collate_examples(train_examples[:2], tokenizer.pad_token_id)
print(
    "SFT tensors [batch_size, padded_sequence_length]:",
    {key: tuple(value.shape) for key, value in preview_batch.items()},
)
boundary_positions = range(prompt_length - 2, prompt_length + 3)
display(
    pd.DataFrame(
        [
            {
                "input_position": position,
                "role": "prompt" if position < prompt_length else "reference response",
                "token_id": example["input_ids"][position],
                "decoded_token": repr(tokenizer.decode([example["input_ids"][position]])),
                "training_label": example["labels"][position],
                "predicting_logits_position": position - 1,
            }
            for position in boundary_positions
        ]
    )
)
for row_index, record in enumerate(train_examples[:2]):
    complete_length = len(record["input_ids"])
    assert (preview_batch["labels"][row_index, : record["prompt_length"]] == -100).all()
    assert preview_batch["labels"][row_index, complete_length - 1].item() == tokenizer.eos_token_id
    assert (preview_batch["labels"][row_index, complete_length:] == -100).all()
print("Prompt masks, response boundaries, real EOS targets, and padding checks passed.")
```

## 4. Understand the learning signal

### 4.1 One answer token versus all possible next choices

Suppose a prefix ends with `3 + 2 =`. The following table uses a
**made-up three-token vocabulary**.

| Next token | Teacher's probability | Student's probability |
| :--- | ---: | ---: |
| `5` | 80% | 10% |
| `6` | 15% | 70% |
| `7` | 5% | 20% |

SFT would say, "the reference token is `5`, so make that token less surprising."
Forward KL instead compares both complete lists. The teacher strongly favors
`5`, while the student favors `6`. Learning should move the student's list toward
the teacher's. This is what **comparing possible next choices** means.

The real models have 151,936 output columns, not three. We keep all columns.
The sampled token's log probability is useful for inspection, but is **not**
the full-vocabulary KL loss.

### 4.2 The equation and the Python operation

For one student-written prefix, let $q$ be the teacher's target distribution and
$p_\theta$ the student's distribution. The temperature $T$ controls how concentrated the teacher target becomes.

$$
q(v)=\operatorname{softmax}(z_T/T)_v,\qquad
p_\theta(v)=\operatorname{softmax}(z_S)_v.
$$

$$
D_{\mathrm{KL}}(q\Vert p_\theta)
=\sum_{v\in\mathcal V} q(v)\big[\log q(v)-\log p_\theta(v)\big].
$$

`forward_kl_per_token` implements this expression directly using FP32
`log_softmax`, multiplication, and a vocabulary sum. A natural logarithm gives
units called **nats**. Zero means identical distributions. Small negative values
near zero can arise from floating-point rounding. Larger negatives are errors.

For a batch of $B$ sampled responses, response $i$ has $R_i$ tokens. The objective
averages positions within each response, then averages those response losses.

$$
\mathcal L_{\mathrm{OPD}}(\theta)
=\frac{1}{B}\sum_{i=1}^{B}\frac{1}{R_i}
\sum_{t=1}^{R_i} D_{\mathrm{KL}}\!\left(
q(\cdot\mid c_T(x_i),y_{i,<t})\Vert p_\theta(\cdot\mid c_S(x_i),y_{i,<t})\right).
$$

Here $c_S(x)$ is the student's original question prompt and $c_T(x)$ is the
teacher's non-thinking chat prompt for that question. The sampled response prefix
$y_{<t}$ is identical for both models. They have different question wrappers and
therefore different prompt lengths. This is an explicit adaptation of GKD.

First average across each response, then across responses, as in GKD's sequence
divergence. A long answer does not outweigh a short answer just by being longer.
Both 04 arms use this per-response weighting. 03 used a pooled token mean, which
is a different valid weighting and remains unchanged there.

The related soft-target cross-entropy is $H(q,p)=-\sum_v q(v)\log p(v)$.
Since $D_{\mathrm{KL}}(q\Vert p)=H(q,p)-H(q)$ and the teacher is frozen,
both have the same student gradient on fixed prefixes. We report **actual KL**,
including the entropy term, rather than calling cross-entropy KL.

### 4.3 Two temperatures, two separate jobs

- **Student sampling temperature = 1.0** draws responses from its current distribution.
  `top_k=0` and `top_p=1.0` remove truncation filters.
- **Teacher target temperature = 0.1** sharpens the teacher's choices for this
  arithmetic experiment. This does not control student sampling. The student
  loss distribution stays at temperature 1, with no extra $T^2$ multiplier.

We do not differentiate through the sampled token IDs. We run a new differentiable
student forward pass on their fixed prefixes. This is the GKD update, not a
policy-gradient estimate through the sampling process.
### 4.4 Two spellings of the same ending

Imagine both models agree there is an 80% chance the answer should end. One puts
that probability on `endoftext`, and the other on `im_end`. Raw-token KL penalizes
that difference. The grouped loss asks whether they agree on the decision to stop.
In this single-turn experiment, both ending IDs are accepted by the decoder.

![Illustration of two ending tokens grouped into one STOP event](assets/04_on_policy_distillation/eos_alignment.png)

*Illustrative probabilities. These numbers are separate from model measurements.*

Let $E$ contain the two ending IDs. Normalize the teacher with its temperature
first, then sum ending probabilities for both models.

$$
q_{\mathrm{stop}}=\sum_{e\in E}q(e),\qquad
p_{\mathrm{stop}}=\sum_{e\in E}p_\theta(e).
$$

The grouped forward KL at every response position is

$$
\sum_{v\notin E}q(v)\log\frac{q(v)}{p_\theta(v)}
+q_{\mathrm{stop}}\log\frac{q_{\mathrm{stop}}}{p_{\mathrm{stop}}}.
$$

`eos_token_ids=None` keeps the raw-token calculation. Passing the verified ending
IDs uses `logsumexp` over normalized log probabilities to calculate STOP mass.
All ordinary vocabulary contributions remain in the loss. We average response
positions and responses exactly as before. Original sampled IDs are preserved.
No synthetic EOS is added to a capped response.

The EOS-alignment paper studies sampled-token updates. This notebook applies its
semantic-action idea to full-distribution forward KL and tests that adaptation.
We report both KL definitions for both pilots because grouping changes the metric.
[EOS-alignment paper](https://arxiv.org/html/2609.20511v1)


```python
# These three probabilities are an ILLUSTRATION, not measurements from Qwen.
# Use teacher temperature 1 here because the table already specifies q directly.
toy_teacher_probs = torch.tensor([[[0.80, 0.15, 0.05]]])  # [1, 1, 3]
toy_student_probs = torch.tensor([[[0.10, 0.70, 0.20]]])
toy_kl = forward_kl_per_token(toy_student_probs.log(), toy_teacher_probs.log(), 1.0)
manual_kl = (toy_teacher_probs * (toy_teacher_probs.log() - toy_student_probs.log())).sum(-1)
assert torch.allclose(toy_kl, manual_kl, atol=1e-6)
assert toy_kl.item() > 0
print(f"Illustration only: teacher-to-student forward KL = {toy_kl.item():.4f} nats.")
print("No model weights were loaded or updated by this example.")
```

```python
# These helpers cache completed evaluations only inside this fixed-signature run.
# They retain every response, token ID, stop flag, and frozen checker result.
def evaluate_and_save(model, examples, name, *, directory=None, identity=None, budget=None):
    # Each completed question is saved immediately. On a new allocation, matching
    # cached rows are checked and only the remaining questions are generated.
    return evaluate_incrementally(
        model,
        tokenizer,
        examples,
        evaluation_policy,
        device,
        compute_dtype,
        path=Path(directory or run_directory) / f"{name}.json",
        identity=f"{identity or signature}:{name}",
        budget=budget or SlurmBudget.from_environment(),
    )


if MODE == "train":
    starting_student = load_student(starting_checkpoint_directory, device)
    starting_validation_loss = validation_loss(
        starting_student, validation_examples, tokenizer.pad_token_id, device, compute_dtype
    )
    starting_validation = evaluate_and_save(
        starting_student, validation_examples, "starting_sft_validation"
    )
    display(pd.DataFrame([evaluation_summary(starting_validation, "starting_sft", "validation")]))
    print("Starting validation reference-token NLL:", starting_validation_loss)
    write_json(
        run_directory / "starting_sft_metrics.json",
        {
            "validation_reference_nll": starting_validation_loss,
            "validation": evaluation_summary(starting_validation, "starting_sft", "validation"),
        },
    )
    # A separate seed and forked RNG make fixed diagnostic responses reproducible
    # without consuming the random stream later used for on-policy training.
    diagnostic_path = run_directory / "fixed_validation_trajectories.json"
    if diagnostic_path.exists():
        diagnostic_trajectories = json.loads(diagnostic_path.read_text())
    else:
        with torch.random.fork_rng(devices=[device.index]):
            torch.manual_seed(config.seed + 10000)
            diagnostic_trajectories = [
                sample_trajectory(
                    starting_student,
                    tokenizer,
                    row,
                    device=device,
                    dtype=compute_dtype,
                    max_new_tokens=128,
                    stop_token_ids=stop_token_ids,
                    policy_step=0,
                )
                for row in validation_examples[: config.diagnostic_examples]
            ]
        write_json(diagnostic_path, diagnostic_trajectories)
    # Fixed validation trajectories are diagnostics only. They never enter training.
    del starting_student
    release_cuda_memory()
else:
    print("INSPECT: no baseline generation, model loss, or GPU performance measured.")
```

## 6. Prepare the comparison and SFT control

The default fresh comparison needs your verified 03 checkpoint and matching report.
Both EOS pilots begin from those same weights. Their shared baseline is measured
in Section 5. This short comparison tests whether grouping ending probabilities
stabilizes learning while preserving task performance.

The earlier Amarel SFT control reached 40/64 strict validation matches. That is
historical evidence. For a later full experiment, this fresh workflow trains a
new 128-update SFT control for a paired evaluation on the reserved test questions.
You can optionally set `HISTORICAL_RUN_DIRECTORY` to verify and reuse the old
control and investigate its saved student checkpoints.

The visible loop below serves both training methods. An optimizer update changes
student weights after eight examples contribute gradients. All eight on-policy
responses come from the same policy version before the update.

Defaults use FP32 student weights, BF16 autocast, gradient checkpointing, AdamW at
`5e-6`, weight decay `0.01`, clipping at `1.0`, and a 128-update linear schedule
with 10% warmup. The first 32 KD updates have checks every eight updates. Each arm
retains its two latest recovery snapshots.


```python
def train_arm(
    arm,
    teacher=None,
    *,
    stop_after_steps=None,
    run_config=None,
    output_directory=None,
    run_signature=None,
    recover_pause_validation=False,
):
    from dataclasses import asdict
    from datetime import datetime, timezone

    from transformers import GenerationConfig

    from src.opd_experiment import (
        SlurmBudget,
        commit_boundary,
        isolated_evaluation_rng,
        publish_pause,
    )
    from src.opd_experiment import (
        WalltimePause as TrainingWalltimePause,
    )

    # Explicit overrides keep comparison arms independent. Existing pilot/full
    # calls use the notebook's selected recipe and run directory.
    config = run_config if run_config is not None else globals()["config"]
    run_directory = (
        Path(output_directory) if output_directory is not None else globals()["run_directory"]
    )
    signature = run_signature if run_signature is not None else globals()["signature"]
    from src.training_safety import (
        accept_checkpoint,
        capture_update_rng,
        pin_starting_checkpoint,
        record_training_failure,
        restore_boundary_measurements,
        restore_update_rng,
        save_boundary_measurements,
    )

    # The new policy qualifies KD pilots. The fixed supervised control keeps
    # its existing recipe and cannot overwrite a KD accepted-checkpoint pointer.
    safety_policy = globals().get("safety_policy") if arm == "on_policy" else None
    baseline_accuracy = (
        globals().get("starting_validation", {}).get("scores", {}).get("strict", {}).get("accuracy")
    )

    def assess_health():
        if safety_policy is not None or arm == "on_policy":
            return training_health(
                history,
                config,
                starting_validation_loss,
                safety_policy=safety_policy,
                baseline_accuracy=baseline_accuracy,
            )
        return {"pause": False, "warnings": []}

    def pause_kind(assessment):
        if any(r["metric"] != "recent_cap_fraction" for r in assessment.get("reasons", [])):
            return "quality_paused"
        return "health_paused"

    budget = SlurmBudget.from_environment()
    assert arm in {"sft_control", "on_policy"}
    if arm == "on_policy" and teacher is None:
        raise ValueError("On-policy KD needs the frozen teacher.")
    target_steps = config.optimizer_steps if stop_after_steps is None else stop_after_steps
    if type(target_steps) is not int or not 1 <= target_steps <= config.optimizer_steps:
        raise ValueError("The requested stage must end within the fixed experiment.")
    final_directory = run_directory / "checkpoints" / arm
    result_path = run_directory / f"{arm}_training.json"
    progress_path = run_directory / "progress.jsonl"

    def record_progress(update, stage, event="started", **details):
        # Close each entry before starting the next operation so it can be
        # read while the run is active or after the kernel is interrupted.
        entry = {
            "timestamp_utc": datetime.now(timezone.utc).isoformat(),
            "arm": arm,
            "update": update,
            "stage": stage,
            "event": event,
            **details,
        }
        with progress_path.open("a", encoding="utf-8") as log:
            log.write(json.dumps(entry) + "\n")

    if final_directory.exists() and result_path.exists():
        verify_inventory(final_directory)
        completed = json.loads(result_path.read_text())
        assert completed["completed_steps"] == config.optimizer_steps
        assert completed.get("signature", signature) == signature
        print(f"{arm}: reusing its verified completed checkpoint.")
        return completed

    snapshots = sorted((run_directory / "resume" / arm).glob("step_*"))
    resume_path = snapshots[-1] if snapshots else None
    resume_metadata = (
        read_resume(resume_path, arm=arm, signature=signature) if resume_path else None
    )
    history = resume_metadata["history"] if resume_metadata else []
    completed_steps = len(history)
    if safety_policy is not None:
        if resume_metadata and resume_metadata.get("training_safety_policy") != asdict(
            safety_policy
        ):
            raise ValueError("Resume safety settings differ. Start a new experiment.")
        pin_starting_checkpoint(
            run_directory, starting_checkpoint_directory, signature, safety_policy
        )
        if resume_path is not None:
            restore_boundary_measurements(
                run_directory, history, resume_path, signature, safety_policy
            )
    if completed_steps > target_steps:
        raise ValueError("This snapshot is beyond the pilot boundary. Choose the full stage.")
    health = assess_health()
    if health["pause"] and not recover_pause_validation:
        raise RuntimeError(
            "This snapshot was health-paused. Review the evidence and start a revised experiment."
        )
    model_source = resume_path / "student" if resume_path else starting_checkpoint_directory
    student = load_student(model_source, device)
    student.generation_config = GenerationConfig.from_dict(evaluation_policy)
    student.generation_config.use_cache = True
    # These fields travel with every newly saved Hugging Face checkpoint.
    student.config.opd_eos_loss_mode = config.eos_loss_mode
    student.config.opd_eos_token_ids = list(config.eos_token_ids)

    def measure_boundary(student, record, name, *, generate_answers=True):
        # A checksummed optimizer snapshot precedes this measurement. Metrics
        # live in an atomic sidecar because the snapshot itself stays immutable.
        boundary_snapshot = run_directory / "resume" / arm / f"step_{record['step']:04d}"
        if safety_policy is not None:
            restore_boundary_measurements(
                run_directory, history, boundary_snapshot, signature, safety_policy
            )
            if "generated_validation" in record:
                return evaluate_and_save(
                    student,
                    validation_examples,
                    name,
                    directory=run_directory,
                    identity=signature,
                    budget=budget,
                )
        # Greedy validation and diagnostic forwards preserve rollout RNG states.
        with isolated_evaluation_rng():
            budget.require_time()
            record_progress(record["step"], "validation")
            record["validation_reference_nll"] = validation_loss(
                student, validation_examples, tokenizer.pad_token_id, device, compute_dtype
            )
            if safety_policy is not None:
                save_boundary_measurements(
                    run_directory,
                    record,
                    boundary_snapshot,
                    signature,
                    safety_policy,
                    complete=False,
                )
                assessment = assess_health()
                if assessment["pause"]:
                    publish_pause(
                        run_directory,
                        arm=arm,
                        status=pause_kind(assessment),
                        step=len(history),
                        snapshot=boundary_snapshot,
                        signature=signature,
                        config=config,
                        health=assessment,
                        reason="validation quality threshold",
                    )
            if teacher is not None:
                budget.require_time()
                diagnostics = fixed_trajectory_diagnostics(
                    student, teacher, diagnostic_trajectories, device, compute_dtype, config
                )
                record["fixed_validation_kl"] = diagnostics["mean_forward_kl_nats"]
                record["fixed_raw_token_kl"] = diagnostics["mean_raw_token_kl_nats"]
                record["fixed_semantic_stop_kl"] = diagnostics["mean_semantic_stop_kl_nats"]
            if not generate_answers:
                record_progress(record["step"], "validation", "completed")
                return None
            measured = evaluate_and_save(
                student,
                validation_examples,
                name,
                directory=run_directory,
                identity=signature,
                budget=budget,
            )
        record["generated_validation"] = evaluation_summary(measured, arm, "validation")
        if safety_policy is not None:
            save_boundary_measurements(
                run_directory, record, boundary_snapshot, signature, safety_policy, complete=True
            )
            if not assess_health()["pause"]:
                accept_checkpoint(
                    run_directory, boundary_snapshot, record, signature, safety_policy
                )
        record_progress(record["step"], "validation", "completed")
        return measured

    def finish_pause(student, status, reason, snapshot):
        # Publish status BEFORE validation. A timeout can leave validation pending
        # while the optimizer boundary and reason remain inspectable on disk.
        health = assess_health()
        result = publish_pause(
            run_directory,
            arm=arm,
            status=status,
            step=len(history),
            snapshot=snapshot,
            signature=signature,
            config=config,
            health=health,
            reason=reason,
        )
        result.update(history=history, resumed_from_step=completed_steps)
        if history:
            try:
                measured = measure_boundary(
                    student, history[-1], f"{arm}_boundary_{len(history):04d}_validation"
                )
                result.update(
                    validation_status="completed",
                    final_validation=evaluation_summary(measured, status, "validation"),
                    final_validation_reference_nll=history[-1]["validation_reference_nll"],
                )
            except TrainingWalltimePause:
                print(
                    "Checkpoint saved; validation will resume in the next allocation.", flush=True
                )
        result["health"] = assess_health()
        if safety_policy is not None and result["health"]["pause"]:
            status = pause_kind(result["health"])
            result["status"] = status
            result["pause_reason"] = "training or validation safety threshold"
        write_json(run_directory / f"{arm}_history.json", history)
        write_json(run_directory / f"{arm}_pause.json", result)
        write_json(run_directory / f"{arm}_{status}.json", result)
        write_json(run_directory / "manifest.json", result)
        # Publish the small report here so a disconnected notebook client does
        # not have to execute the later export cell to retain this boundary.
        if safety_policy is not None:
            try:
                export_report(
                    PROJECT_ROOT,
                    run_directory,
                    result,
                    report_name=f"report_{status}_{len(history):04d}_{time.time_ns()}",
                )
            except OSError as error:
                # The checkpoint and pause manifest already exist. A full disk
                # or transfer error must remain visible without hiding that state.
                print(
                    f"Report packaging failed. Checkpoint retained at {snapshot}. {error}",
                    flush=True,
                )
        return result

    trajectory_path = run_directory / "training_trajectories.jsonl"
    if arm == "on_policy":
        truncate_uncommitted_trajectories(trajectory_path, completed_steps)
        validate_trajectory_log(
            trajectory_path, train_examples, completed_steps, config.accumulation_steps
        )

    # A comparison may finish interrupted evaluation of an already stopped arm.
    # It never restarts updates from a health-paused policy.
    if recover_pause_validation and (health["pause"] or completed_steps == target_steps):
        try:
            return finish_pause(
                student,
                pause_kind(health) if health["pause"] else "pilot_complete",
                "health guard" if health["pause"] else "pilot boundary",
                resume_path,
            )
        finally:
            del student
            release_cuda_memory()

    seed_everything(config.seed)
    optimizer = torch.optim.AdamW(
        student.parameters(), lr=config.learning_rate, weight_decay=config.weight_decay
    )
    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=math.ceil(config.optimizer_steps * config.warmup_fraction),
        num_training_steps=config.optimizer_steps,
    )
    scaler = torch.amp.GradScaler("cuda", enabled=False)
    if resume_path:
        restore_training_state(resume_path, optimizer, scheduler, scaler)
    if teacher is not None:
        optimizer_ids = {id(p) for group in optimizer.param_groups for p in group["params"]}
        assert not optimizer_ids.intersection(id(p) for p in teacher.parameters())
        assert all(not p.requires_grad and p.grad is None for p in teacher.parameters())
    optimizer.zero_grad(set_to_none=True)
    if torch.device(device).type == "cuda":
        for index in range(torch.cuda.device_count()):
            torch.cuda.reset_peak_memory_stats(index)
        torch.cuda.synchronize()
    started = time.perf_counter()
    pause_status, pause_reason = None, None
    latest_snapshot = resume_path
    update_pending = False
    update_rng = None

    def save_boundary(student, optimizer, scheduler, scaler, *, status=None, reason=None):
        # At step zero, the verified 03 checkpoint already supplies the start state.
        snapshot = None
        if history:
            record_progress(len(history), "checkpoint")
            snapshot = commit_boundary(
                run_directory,
                student,
                tokenizer,
                optimizer,
                scheduler,
                scaler,
                arm=arm,
                step=len(history),
                signature=signature,
                history=history,
                config=config,
                training_safety_policy=safety_policy,
            )
        if status is not None:
            # Publish immediately after the immutable checkpoint becomes visible.
            # Cleanup or later validation can then be interrupted safely.
            publish_pause(
                run_directory,
                arm=arm,
                status=status,
                step=len(history),
                snapshot=snapshot,
                signature=signature,
                config=config,
                health=health,
                reason=reason,
            )
        if history:
            record_progress(len(history), "checkpoint", "completed")
        return snapshot

    try:
        # A boundary snapshot can precede its validation. Finish that work before
        # generating the next group, using cached answers from interrupted evaluation.
        if history and "generated_validation" not in history[-1]:
            measure_boundary(student, history[-1], f"{arm}_boundary_{len(history):04d}_validation")
        for update_index in range(completed_steps + 1, target_steps + 1):
            # Re-evaluate recovered validation BEFORE allowing another update.
            health = assess_health()
            if health["pause"]:
                pause_status, pause_reason = (
                    pause_kind(health),
                    "saved validation or stopping threshold",
                )
                latest_snapshot = save_boundary(
                    student, optimizer, scheduler, scaler, status=pause_status, reason=pause_reason
                )
                break
            budget.require_time()
            # If the deadline arrives mid-group, restore the RNG from this
            # clean boundary before saving the previous completed update.
            update_rng = capture_update_rng() if safety_policy is not None else None
            update_pending = True
            group_start = (update_index - 1) * config.accumulation_steps
            examples = train_examples[group_start : group_start + config.accumulation_steps]
            step_started = time.perf_counter()
            student.train()
            if arm == "on_policy":
                # All eight fresh responses come from the same unchanged policy.
                record_progress(update_index, "student_generation")
                trajectories = []
                for example_index, example in enumerate(examples):
                    if safety_policy is not None:
                        budget.require_time()
                        record_progress(
                            update_index,
                            "student_generation",
                            example_index=example_index,
                            example_id=example["example_id"],
                        )
                    trajectories.append(
                        sample_trajectory(
                            student,
                            tokenizer,
                            example,
                            device=device,
                            dtype=compute_dtype,
                            max_new_tokens=config.max_response_length,
                            stop_token_ids=stop_token_ids,
                            policy_step=update_index - 1,
                        )
                    )
                record_progress(update_index, "teacher_scoring_and_backward")
                metrics = backward_opd_group(
                    student,
                    teacher,
                    trajectories,
                    device,
                    compute_dtype,
                    config,
                    scaler,
                    policy_step=update_index - 1,
                    before_trajectory=budget.require_time if safety_policy is not None else None,
                )
            else:
                # Reference response labels receive cross-entropy. Prompt/padding
                # labels are -100, preserving real EOS tokens as training targets.
                record_progress(update_index, "supervised_forward_and_backward")
                metrics = backward_sft_group(
                    student, examples, tokenizer.pad_token_id, device, compute_dtype, scaler
                )
            record_progress(update_index, "optimizer_step")
            update_metrics = optimizer_update(
                student, optimizer, scheduler, scaler, config.max_grad_norm
            )
            if torch.device(device).type == "cuda":
                torch.cuda.synchronize()
            record = {
                "step": update_index,
                **metrics,
                **update_metrics,
                "update_seconds": time.perf_counter() - step_started,
                "peak_allocated_gpu_gib_so_far": torch.cuda.max_memory_allocated() / 2**30
                if torch.device(device).type == "cuda"
                else None,
            }
            if torch.device(device).type == "cuda":
                record["gpu_memory"] = [
                    {
                        "index": index,
                        "peak_allocated_gib": torch.cuda.max_memory_allocated(index) / 2**30,
                    }
                    for index in range(torch.cuda.device_count())
                ]
            if arm == "on_policy":
                append_trajectories(trajectory_path, trajectories)
                del trajectories
            history.append(record)
            update_pending = False
            health = assess_health()
            for warning in health["warnings"]:
                print("REVIEW:", warning, flush=True)
            write_json(run_directory / f"{arm}_history.json", history)
            record_progress(update_index, "training_update", "completed")
            print(
                f"{arm} [{config.eos_loss_mode}] | update {update_index:3d}/{config.optimizer_steps} | "
                f"loss {metrics['loss']:.4f} | mean response tokens {metrics['response_tokens'] / metrics['examples']:.1f} | "
                f"cap stops {metrics['token_limit_stops']}/{metrics['examples']} | EOS stops {metrics.get('eos_stops', 'reference')}",
                flush=True,
            )
            interval = (
                config.pilot_check_interval
                if arm == "on_policy" and update_index <= config.pilot_updates
                else config.checkpoint_interval
            )
            validation_interval = (
                config.pilot_check_interval
                if update_index <= config.pilot_updates
                else config.validation_interval
            )
            boundary = (
                update_index % interval == 0 or update_index == target_steps or health["pause"]
            )
            if health["pause"]:
                pause_status, pause_reason = pause_kind(health), "training safety threshold"
            elif update_index == target_steps and target_steps < config.optimizer_steps:
                pause_status, pause_reason = "pilot_complete", "pilot boundary"
            if boundary:
                latest_snapshot = save_boundary(
                    student, optimizer, scheduler, scaler, status=pause_status, reason=pause_reason
                )
            if pause_status is not None:
                # finish_pause saves each follow-up answer incrementally.
                break
            if update_index % validation_interval == 0 or update_index == target_steps:
                # Save BEFORE the potentially slow generation of validation answers.
                if not boundary:
                    latest_snapshot = save_boundary(student, optimizer, scheduler, scaler)
                measure_boundary(
                    student,
                    record,
                    f"{arm}_boundary_{update_index:04d}_validation",
                    generate_answers=arm == "on_policy" or update_index == target_steps,
                )
                write_json(run_directory / f"{arm}_history.json", history)
                # Fresh validation can pause immediately, before the next update.
                health = assess_health()
                if health["pause"]:
                    pause_status, pause_reason = pause_kind(health), "validation quality threshold"
                    latest_snapshot = save_boundary(
                        student,
                        optimizer,
                        scheduler,
                        scaler,
                        status=pause_status,
                        reason=pause_reason,
                    )
                    break
    except TrainingWalltimePause:
        if update_pending and update_rng is not None:
            optimizer.zero_grad(set_to_none=True)
            restore_update_rng(update_rng)
        pause_status, pause_reason = "walltime_paused", "SLURM checkpoint reserve"
        latest_snapshot = save_boundary(
            student, optimizer, scheduler, scaler, status=pause_status, reason=pause_reason
        )
    except (KeyboardInterrupt, RuntimeError, FloatingPointError, OSError) as error:
        # An incomplete update is never published as a valid recovery snapshot.
        # Existing complete boundaries remain available even if storage is full.
        try:
            failure = record_training_failure(
                run_directory,
                error=error,
                step=len(history),
                snapshot=latest_snapshot,
                signature=signature,
            )
            if safety_policy is not None:
                export_report(
                    PROJECT_ROOT,
                    run_directory,
                    failure,
                    report_name=f"report_failure_{len(history):04d}_{time.time_ns()}",
                )
        except (OSError, ValueError) as export_error:
            print("Failure-report export could not finish:", export_error, flush=True)
        del optimizer, scheduler, scaler, student
        release_cuda_memory()
        raise
    elapsed = time.perf_counter() - started
    peak_gib = (
        torch.cuda.max_memory_allocated() / 2**30 if torch.device(device).type == "cuda" else None
    )
    del optimizer, scheduler, scaler
    release_cuda_memory()
    if arm == "on_policy":
        validate_trajectory_log(
            trajectory_path, train_examples, len(history), config.accumulation_steps
        )
    if pause_status is not None or len(history) < config.optimizer_steps:
        status = pause_status or "pilot_complete"
        try:
            return finish_pause(student, status, pause_reason or "pilot boundary", latest_snapshot)
        finally:
            del student
            release_cuda_memory()

    record_progress(len(history), "final_checkpoint")
    final_directory.parent.mkdir(parents=True, exist_ok=True)
    if final_directory.exists():
        verify_inventory(final_directory)
    else:
        save_checkpoint(student, tokenizer, final_directory)
    checkpoint_verified = verify_checkpoint(
        student,
        tokenizer,
        final_directory,
        validation_examples[0]["input_ids"],
        device,
        compute_dtype,
        probe_texts=["  spaces\n", "3 + 2 = 5.\n#### 5", "Café 中文 🙂"],
    )
    record_progress(len(history), "final_checkpoint", "completed")
    record_progress(len(history), "validation")
    final_validation = evaluate_and_save(
        student,
        validation_examples,
        f"{arm}_boundary_{len(history):04d}_validation",
        directory=run_directory,
        identity=signature,
        budget=budget,
    )
    result = {
        "arm": arm,
        "status": "completed",
        "signature": signature,
        "eos_loss_mode": config.eos_loss_mode,
        "validation_status": "completed",
        "health": assess_health(),
        "completed_steps": len(history),
        "history": history,
        "current_session_loop_seconds": elapsed,
        "peak_allocated_gpu_gib": peak_gib,
        "summed_recorded_update_seconds": sum(row["update_seconds"] for row in history),
        "maximum_recorded_peak_gpu_gib": max(
            (
                row["peak_allocated_gpu_gib_so_far"]
                for row in history
                if row["peak_allocated_gpu_gib_so_far"] is not None
            ),
            default=None,
        ),
        "checkpoint_reload_verified": checkpoint_verified,
        "final_validation": evaluation_summary(final_validation, arm, "validation"),
        "final_validation_reference_nll": history[-1]["validation_reference_nll"],
        "resumed_from_step": completed_steps,
    }
    write_json(result_path, result)
    record_progress(len(history), "validation", "completed")
    del student
    release_cuda_memory()
    return result
```

## 7. Qualify the teacher and prove response alignment

The teacher receives its non-thinking chat question context, followed by the
**exact response IDs the student generated**. Its prefix length is $P_T$, while the
student's is $P_S$. Their first-response predictions occur at $P_T-1$ and $P_S-1$,
respectively. The aligned response-logit tensors both have shape $[1,R,V]$ even
when their complete-sequence lengths differ.

For example, the same response IDs for `3 + 2 = 5` follow a plain `Answer:` prefix
for the student and an assistant-turn prefix for the teacher. We append those IDs
directly, avoiding changed token boundaries from decoding and retokenizing them.

We also generate teacher answers on the **64 validation questions** to measure
its format, numeric matching, and stopping behavior under this context. These
answers are never substituted into distillation training. Greater model size and
higher stopping probability alone do not prove useful arithmetic supervision.

`eval()` selects inference behavior. `requires_grad_(False)` freezes parameters,
and `torch.no_grad()` prevents a teacher gradient graph. KV cache is disabled for
scoring known sequences and enabled for autoregressive evaluation.


```python
if MODE == "train":
    print(
        "Loading the frozen 32B teacher, about 61 GiB of weights. The first download can take several minutes.",
        flush=True,
    )
    teacher = load_teacher(device)
    # The frozen teacher may span two GPUs. Never call teacher.to(device),
    # which would collapse its layer placement onto one card.
    runtime["teacher_device_map"] = {
        str(k): str(v) for k, v in getattr(teacher, "hf_device_map", {"": str(device)}).items()
    }
    # Teacher-generated answers are an independent validation sanity check.
    teacher_validation_examples = []
    for row in validation_examples:
        teacher_prompt = teacher_tokenizer.decode(
            row["teacher_prompt_token_ids"], skip_special_tokens=False
        )
        assert (
            tokenizer.encode(teacher_prompt, add_special_tokens=False)
            == row["teacher_prompt_token_ids"]
        )
        teacher_validation_examples.append({**row, "prompt": teacher_prompt})
    teacher_validation = evaluate_and_save(
        teacher, teacher_validation_examples, "teacher_validation"
    )
    display(pd.DataFrame([evaluation_summary(teacher_validation, "teacher", "validation")]))
    if (
        teacher_validation["scores"]["strict"]["accuracy"]
        <= starting_validation["scores"]["strict"]["accuracy"]
    ):
        print("REVIEW: the teacher did not exceed starting-student strict validation accuracy.")
    student = load_student(starting_checkpoint_directory, device)
    student.eval()
    trajectory = diagnostic_trajectories[0]
    complete_sequence_ids, attention_mask = trajectory_tensors(trajectory, device)
    teacher_sequence_ids, teacher_attention_mask = trajectory_tensors(
        trajectory, device, context="teacher"
    )
    prompt_length = trajectory["prompt_length"]
    teacher_prompt_length = trajectory["teacher_prompt_length"]
    response_token_ids = complete_sequence_ids[:, prompt_length:]
    response_length = response_token_ids.shape[1]
    # Each model has its own [1, complete_length] input and causal prompt offset.
    with torch.no_grad(), precision_context(device, compute_dtype):
        student_logits = student(
            input_ids=complete_sequence_ids, attention_mask=attention_mask, use_cache=False
        ).logits
        teacher_logits = teacher(
            input_ids=teacher_sequence_ids, attention_mask=teacher_attention_mask, use_cache=False
        ).logits
    assert teacher_logits.shape[:2] == teacher_sequence_ids.shape
    assert student_logits.shape[:2] == complete_sequence_ids.shape
    print(
        "Student logits [batch, student_sequence, vocabulary]:",
        tuple(student_logits.shape),
    )
    student_response_logits = response_prediction_logits(
        student_logits, complete_sequence_ids, prompt_length
    )
    teacher_response_logits = response_prediction_logits(
        teacher_logits, teacher_sequence_ids, teacher_prompt_length
    ).to(device, copy=True)
    assert student_response_logits.shape == teacher_response_logits.shape
    assert torch.equal(
        complete_sequence_ids[:, prompt_length:], teacher_sequence_ids[:, teacher_prompt_length:]
    )
    print("Teacher logits [batch, teacher_sequence, vocabulary]:", tuple(teacher_logits.shape))
    print(
        "Aligned response logits [batch, response_tokens, vocabulary]:",
        tuple(student_response_logits.shape),
    )
    # These selected scores describe the tokens that appeared. The KL below
    # separately compares ALL vocabulary choices at each response position.
    response_student_log_probs = selected_log_probs(student_response_logits, response_token_ids)
    response_teacher_target_log_probs = selected_log_probs(
        teacher_response_logits.float() / config.teacher_temperature, response_token_ids
    )
    assert (
        response_student_log_probs.shape
        == response_teacher_target_log_probs.shape
        == (1, response_length)
    )
    assert torch.equal(complete_sequence_ids[:, 1:][:, prompt_length - 1 :], response_token_ids)
    assert (
        torch.isfinite(student_response_logits).all()
        and torch.isfinite(teacher_response_logits).all()
    )
    assert (
        torch.isfinite(response_student_log_probs).all() and (response_student_log_probs <= 0).all()
    )
    assert (
        torch.isfinite(response_teacher_target_log_probs).all()
        and (response_teacher_target_log_probs <= 0).all()
    )
    smoke_kl = forward_kl_loss(
        student_response_logits,
        teacher_response_logits,
        teacher_temperature=config.teacher_temperature,
        chunk_size=config.kl_chunk_size,
        recompute=False,
    )
    print("One unchanged diagnostic response: raw-token forward KL =", float(smoke_kl), "nats")
    display(
        pd.DataFrame(
            [
                {
                    "final_prompt_token": repr(
                        tokenizer.decode([complete_sequence_ids[0, prompt_length - 1].item()])
                    ),
                    "first_response_token": repr(
                        tokenizer.decode([response_token_ids[0, 0].item()])
                    ),
                    "first_response_input_position": prompt_length,
                    "student_predicting_position": prompt_length - 1,
                    "teacher_predicting_position": teacher_prompt_length - 1,
                    "student_log_probability": response_student_log_probs[0, 0].item(),
                    "sharpened_teacher_target_log_probability": response_teacher_target_log_probs[
                        0, 0
                    ].item(),
                }
            ]
        )
    )
    display(
        pd.DataFrame(
            [
                {
                    "response_position": position,
                    "token_id": token_id,
                    "decoded_token": repr(tokenizer.decode([token_id])),
                    "student_log_probability": response_student_log_probs[0, position].item(),
                    "sharpened_teacher_target_log_probability": response_teacher_target_log_probs[
                        0, position
                    ].item(),
                }
                for position, token_id in enumerate(trajectory["response_token_ids"][:12])
            ]
        )
    )
    # At the final sampled position, inspect total ending mass across BOTH
    # recognized stop IDs. The teacher can prefer im_end over endoftext.
    # If this response was capped, this position is not a natural end boundary.
    boundary_record = {
        "sampled_ended_with_eos": trajectory["ended_with_eos"],
        "last_sampled_token": repr(tokenizer.decode([trajectory["response_token_ids"][-1]])),
        "student_stop_probability_T1": student_response_logits[0, -1]
        .float()
        .softmax(-1)[stop_token_ids]
        .sum()
        .item(),
        "teacher_stop_probability_T1": teacher_response_logits[0, -1]
        .float()
        .softmax(-1)[stop_token_ids]
        .sum()
        .item(),
        "teacher_target_stop_probability": (
            teacher_response_logits[0, -1].float() / config.teacher_temperature
        )
        .softmax(-1)[stop_token_ids]
        .sum()
        .item(),
    }
    display(pd.DataFrame([boundary_record]))
    write_json(run_directory / "teacher_boundary_probe.json", boundary_record)
    assert all(not p.requires_grad and p.grad is None for p in teacher.parameters())
    print(
        "Frozen teacher, identical response prefixes, separate causal shifts, and finite-value checks passed."
    )
    # Free the inspection tensors before training; retaining a notebook variable
    # can otherwise keep its large vocabulary tensor on the GPU.
    del student, student_logits, teacher_logits, student_response_logits, teacher_response_logits
    del response_student_log_probs, response_teacher_target_log_probs, smoke_kl
    release_cuda_memory()
    diagnostic_student = load_student(starting_checkpoint_directory, device)
    starting_fixed_diagnostics = fixed_trajectory_diagnostics(
        diagnostic_student, teacher, diagnostic_trajectories, device, compute_dtype, config
    )
    write_json(run_directory / "starting_sft_fixed_diagnostics.json", starting_fixed_diagnostics)
    del diagnostic_student
    release_cuda_memory()
else:
    print("INSPECT: teacher weights and real alignment logits still require a CUDA GPU run.")
```

### 7.1 Inspect stopping probabilities before training

A fresh Colab run probes the starting 03 student on its eight fixed validation
responses. For each response, we inspect individual EOS probabilities and their
combined stopping probability. We also compute how each loss would change detached
logits, while keeping model weights fixed.

A prefix containing $k$ response tokens predicts its next token at position
$P+k-1$. Before an observed EOS, that EOS is a target. After a capped response,
there is no observed next token. The probe uses the original response token IDs.

If `HISTORICAL_RUN_DIRECTORY` points to a complete old Amarel run, this section
also probes updates 16 and 22 on identical prefixes and recovers update-22
validation. The default fresh comparison begins with the 03 probe only.

`stopping_probes.json` contains every EOS probability, logit gradient, causal
position, and decoded prefix tail. The preview shows a subset of those records.


```python
# Frozen probes inspect the starting student before any comparison updates.
if MODE == "train" and RUN_STAGE == "eos_comparison":
    if HISTORICAL_RUN_DIRECTORY is None:
        # Reuse original response IDs from the starting student's fixed validation
        # trajectories. The fresh run can diagnose EOS probabilities before training.
        selected_probes = [
            dict(row, probe_group="fresh_starting_validation") for row in diagnostic_trajectories
        ]
        probe_checkpoints = {"starting_03": starting_checkpoint_directory}
    else:
        selected_probes = load_historical_probes(HISTORICAL_RUN_DIRECTORY)
        probe_checkpoints = {
            "starting_03": starting_checkpoint_directory,
            "step_0016": Path(historical_evidence["snapshots"]["16"]) / "student",
            "step_0022": Path(historical_evidence["snapshots"]["22"]) / "student",
        }
    write_json(run_directory / "probe_trajectories.json", selected_probes)
    probe_path = run_directory / "stopping_probes.json"
    probe_identity = hashlib.sha256(
        json.dumps(
            {
                "signature": signature,
                "historical_signature": historical_evidence.get("signature"),
                "original_trajectories": selected_probes,
            },
            sort_keys=True,
        ).encode()
    ).hexdigest()
    if probe_path.exists():
        stopping_probes = json.loads(probe_path.read_text())
        assert stopping_probes["probe_identity"] == probe_identity, (
            "Historical prefixes or checkpoint identities changed. Start a fresh diagnosis."
        )
    else:
        stopping_probes = probe_checkpoint_set(
            teacher,
            tokenizer,
            selected_probes,
            probe_checkpoints,
            device,
            compute_dtype,
            config,
            budget=SlurmBudget.from_environment(),
        )
        stopping_probes["probe_identity"] = probe_identity
        write_json(probe_path, stopping_probes)
    # Group matching prefixes when historical checkpoints are supplied.
    # A fresh run displays the starting 03 student. The full JSON retains
    # every individual EOS probability, logit gradient, and decoded prefix tail.
    probe_table = pd.DataFrame(stopping_probes["rows"]).sort_values(
        ["probe_group", "example_id", "response_prefix_length", "checkpoint"]
    )
    display(
        probe_table[
            [
                "checkpoint",
                "example_id",
                "boundaries",
                "response_prefix_length",
                "student_predicting_position",
                "teacher_predicting_position",
                "observed_next_token",
                "student_stop_probability",
                "teacher_T1_stop_probability",
                "teacher_target_stop_probability",
                "raw_token_kl_nats",
                "semantic_stop_kl_nats",
            ]
        ].head(18)
    )
    # New probes expose target sharpness and separate EOS contributions.
    # Historical probe files remain readable when these fields were not recorded.
    extra_columns = [
        "teacher_T1_entropy_nats",
        "teacher_target_entropy_nats",
        "ordinary_token_contribution_nats",
        "raw_eos_contribution_nats",
        "grouped_eos_contribution_nats",
    ]
    if all(column in probe_table.columns for column in extra_columns):
        display(probe_table[["example_id", "response_prefix_length", *extra_columns]].head(8))
    if HISTORICAL_RUN_DIRECTORY is not None:
        # The previous allocation ended during this validation. Continue generation
        # from the saved weights, saving each answer into the NEW diagnosis folder.
        recovered_student = load_student(probe_checkpoints["step_0022"], device)
        try:
            recovered_validation = evaluate_and_save(
                recovered_student,
                validation_examples,
                "historical_step_0022_validation",
                identity=f"{historical_evidence['signature']}:step_0022",
            )
            with isolated_evaluation_rng():
                recovered_nll = validation_loss(
                    recovered_student,
                    validation_examples,
                    tokenizer.pad_token_id,
                    device,
                    compute_dtype,
                )
            write_json(
                run_directory / "historical_step_0022_metrics.json",
                {
                    "validation": evaluation_summary(
                        recovered_validation, "historical_step_0022", "validation"
                    ),
                    "reference_nll": recovered_nll,
                },
            )
        finally:
            del recovered_student
            release_cuda_memory()
    print("Stopping probes saved. Review their EOS probabilities and gradients.")
else:
    print("INSPECT: frozen-model stopping probes run on the allocated GPU.")
```

## 8. Compare raw-token and grouped-STOP pilots

**Saved outputs below describe the earlier guard policy.** New executions use
the versioned safeguards printed in Section 2. Their GPU behavior awaits validation.

Both pilots start from the verified 03 weights, use the same training questions
and initial seed, and keep the same 128-update learning-rate schedule. Each is
limited to 32 updates. Their responses can diverge as their weights change.
The frozen teacher is shared. Each student has its own optimizer, random stream,
signature, and recovery folder.

| Setting | Both pilots |
| :--- | :--- |
| Student sampling | Temperature 1, no top-k/top-p filtering |
| Teacher targets | Temperature 0.1, non-thinking chat context |
| Optimizer | AdamW, peak learning rate `5e-6`, clip norm 1 |
| Accumulation | Eight fresh student responses per update |
| Generation limit | 256 response tokens |
| Comparison | Raw-token KL versus semantic STOP KL |

Every eight updates, we save an optimizer-boundary snapshot before validation.
Validation includes reference NLL, generated answers, and both KL definitions on
the same fixed prefixes. The guarded policy pauses above 8 capped responses in a window of
32 and applies the baseline-relative quality limits from Section 2. Pause status is written before generation of the follow-up validation answers.
A health-paused optimizer stays stopped. Its unfinished validation can be recovered.
Each pilot also appends to `progress.jsonl` before each training stage starts.
If the kernel stops mid-update, check the last entry for the operation it reached.

The review gates are deliberately explicit. The grouped pilot should reach 32
updates with finite calculations, at most 8 caps in its final 32 responses,
reference NLL no more than 10% above its starting value, and at least 32/64 strict
validation matches. A meaningful EOS contrast requires at least a 25-percentage-point
cap-rate reduction at the latest common update. These are engineering thresholds
for review, not statistical evidence of generalization.

The primary test remains reserved. If the gates fail or the contrast is unclear,
report that outcome and reconsider the hypothesis before extending training.


A complete checkpoint is saved before validation. The last acceptable checkpoint
is protected from cleanup while the two newest recovery states are retained.
Completed validation measurements are bound to that checkpoint's inventory hash.
A paused run can finish its evaluation before review. Continuing its optimizer
requires a matching accepted policy and an eligible recovery state.


```python
# The two pilot policies start independently from the same verified 03 weights.
# Only EOS loss interpretation differs. The frozen teacher is loaded once.
FULL_RUN_COMPLETE = False
if MODE == "train" and RUN_STAGE == "eos_comparison":
    assert (run_directory / "stopping_probes.json").is_file(), "Run Section 7.1 first."
    comparison_storage = check_comparison_storage(run_directory)
    print("Storage check:", comparison_storage)
    comparison_result = run_eos_comparison(
        train_arm,
        teacher,
        directory=run_directory,
        config=config,
        protocol=protocol,
        splits=splits,
        checkpoint_hashes=checkpoint_hashes,
        evaluation_policy=evaluation_policy,
        code_hashes=code_hashes,
        starting_validation_nll=starting_validation_loss,
        safety_policy=safety_policy,
        report_root=PROJECT_ROOT,
        starting_validation_accuracy=starting_validation["scores"]["strict"]["accuracy"],
    )
    display(
        pd.DataFrame(
            [
                {
                    "eos_loss_mode": mode,
                    "status": result["status"],
                    "completed_updates": result["completed_steps"],
                    "validation_status": result["validation_status"],
                    **result.get("final_validation", {}),
                }
                for mode, result in comparison_result["results"].items()
            ]
        )
    )
    print("Engineering gates:", comparison_result["gates"])
    from IPython.display import Image

    from src.opd_plots import plot_eos_comparison

    plot_eos_comparison(comparison_result["results"], starting_validation_loss, run_directory)
    display(Image(filename=str(run_directory / "eos_comparison_diagnostics.png")))
    print("Review both EOS modes, response text, and task results before a full run.")
elif MODE == "train":
    if RUN_STAGE == "full":
        snapshots = sorted((run_directory / "resume/on_policy").glob("step_*"))
        assert snapshots, "A full stage must resume a reviewed pilot snapshot."
        pilot_metadata = read_resume(snapshots[-1], arm="on_policy", signature=signature)
        assert pilot_metadata["completed_steps"] >= min(
            config.pilot_updates, config.optimizer_steps
        )
    stop_after = min(config.pilot_updates, config.optimizer_steps) if RUN_STAGE == "pilot" else None
    on_policy_result = train_arm("on_policy", teacher=teacher, stop_after_steps=stop_after)
    assert all(not p.requires_grad and p.grad is None for p in teacher.parameters())
    if on_policy_result.get("validation_status") == "completed":
        display(pd.DataFrame([on_policy_result["final_validation"]]))
    print("Run status:", on_policy_result["status"])
    print("Health review:", on_policy_result["health"])
    FULL_RUN_COMPLETE = on_policy_result["status"] == "completed"
else:
    print("INSPECT: both EOS objectives and comparison wiring are ready for GPU qualification.")
```

## 9. Evaluate fixed final checkpoints

The EOS comparison skips this section's test-generation cells. The primary held-out
exam is opened only after a fixed 128-update run completes. Each checkpoint receives the
same questions, prompt, greedy decoding, end-token rules, token budget, and frozen
answer checkers. References are used **after generation** to score the final number.

- **Strict match** counts a correct numeric answer on the required `#### <number>` final line.
- **Format compliance** counts a parseable final line, whether its number is right or wrong.
- **Exploratory prose match** uses the unchanged broader extractor to find a matching final number.
- **Token-limit stops** count responses that exhausted the budget without a recognized end token.

We keep every question in the denominator. A higher format score alone is not
better arithmetic. Numeric agreement also does not establish valid reasoning.
Paired bootstrap intervals summarize uncertainty across these questions, conditional
on one training seed. They do not measure run-to-run training variability.

The previously inspected 32-question subset is **secondary historical context**.
Its freshly generated 04 scores use 04's stopping rule and should not be presented
as a controlled comparison against old 02/03 aggregates.

### 9.1 Read answers that changed

Aggregate scores can hide regressions. The next table includes both
incorrect-to-correct and correct-to-incorrect changes. Read their actual text
before deciding whether the model learned arithmetic, formatting, or a different
way of ending its answer. An empty table can be a valid result.


```python
if MODE == "train" and FULL_RUN_COMPLETE:
    before = primary_evaluations["starting_sft"]
    after = primary_evaluations["on_policy"]
    before_rows = {r["example_id"]: r for r in before["rows"]}
    after_rows = {r["example_id"]: r for r in after["rows"]}
    before_scores = {r["example_id"]: r for r in before["annotations"]}
    changed_answers = []
    for row in after["annotations"]:
        key = row["example_id"]
        if before_scores[key]["strict_correct"] != row["strict_correct"]:
            changed_answers.append(
                {
                    "example_id": key,
                    "change": "improved" if row["strict_correct"] else "regressed",
                    "question": before_rows[key]["question"],
                    "expected_answer": before_rows[key]["expected_answer"],
                    "starting_response": before_rows[key]["response_text"],
                    "on_policy_response": after_rows[key]["response_text"],
                }
            )
    # Include both directions in the visible preview instead of cherry-picking gains.
    preview = [
        r
        for direction in ("improved", "regressed")
        for r in [v for v in changed_answers if v["change"] == direction][:2]
    ]
    display(pd.DataFrame(preview))
    write_json(run_directory / "changed_answers.json", changed_answers)
else:
    print("PENDING: no generated-answer changes to inspect.")
```
